```python
#Data processing and analysis:
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def parse_values(text):
    values = np.array([
        int(line.split("->")[1])
        for line in text.strip().splitlines()
        if "->" in line
    ])[1:]  # eerste waarde weg: foute T
    med = np.median(values)
    return values[np.abs(values - med) < 0.05 * med]  # uitschieters (halve periodes) weg

T_tot = np.array([np.mean(parse_values(s)) for s in [s1, s2, s3, s4]]) / 1e6  # µs -> s
L_tot = np.array([l1, l2, l3, l4])

def trillingstijdformule(l, g):
    return 2 * np.pi * np.sqrt(l / g)

val, cov = curve_fit(trillingstijdformule, L_tot, T_tot, p0=[9.81])
g_fit = val[0]
g_err = np.sqrt(cov[0, 0])

test = np.linspace(min(L_tot), max(L_tot), 1000)
plt.plot(L_tot, T_tot, "o", label="Gemeten")
plt.plot(test, trillingstijdformule(test, g_fit), "-", label=f"Fit: g = {g_fit:.2f} ± {g_err:.2f} m/s²")
plt.plot(test, trillingstijdformule(test, 9.81), "--", label="Theorie (g = 9,81)")
plt.xlabel("l (m)")
plt.ylabel("T (s)")
plt.legend()
plt.show()

print(g_fit, g_err)
```
